```python
import gradio as gr
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch
from gtts import gTTS
import tempfile

# Load BLIP model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

def generate_caption(image):
    image = image.convert("RGB")
    inputs = processor(images=image, return_tensors="pt")
    with torch.no_grad():
        outputs = model.generate(**inputs)
    caption = processor.decode(outputs[0], skip_special_tokens=True)
    return caption

def speak_caption(caption):
    tts = gTTS(text=caption)
    with tempfile.NamedTemporaryFile(delete=False, suffix=".mp3") as fp:
        tts.save(fp.name)
        return fp.name

def save_caption_file(caption):
    with tempfile.NamedTemporaryFile(delete=False, suffix=".txt", mode='w') as f:
        f.write(caption)
        return f.name

# Custom CSS
custom_css = """
body {
    font-family: 'Segoe UI', sans-serif;
}
.gradio-container {
    max-width: 900px !important;
    margin: auto;
    padding: 2rem;
}
.dark .gr-block {
    background-color: #1e293b !important;
    color: white !important;
}
.dark textarea, .dark input, .dark .gr-button {
    background-color: #334155 !important;
    color: white !important;
}
button {
    background-color: #3b82f6 !important;
    color: white !important;
    border-radius: 10px !important;
}
button:hover {
    background-color: #2563eb !important;
}
input[type="file"] {
    border: 2px dashed #3b82f6 !important;
    background-color: #f0f9ff !important;
    border-radius: 10px !important;
}
textarea {
    background-color: #f0f9ff !important;
    color: #1e293b !important;
    font-weight: 500;
    border-radius: 10px !important;
    padding: 10px;
}
"""

with gr.Blocks(css=custom_css, theme=gr.themes.Soft()) as demo:
    gr.Markdown("""
    #  **Image Captioning using ML**
    Upload an image and generate a caption.
    """)

    with gr.Row():
        with gr.Column():
            image_input = gr.Image(type="pil", label="📷 Upload Image")
            caption_output = gr.Textbox(label="📝 Generated Caption", lines=3)

            with gr.Row():
                gen_btn = gr.Button("✨ Generate Caption")
                tts_btn = gr.Button("🔊 Speak")
                download_btn = gr.Button("📥 Download Caption")

        with gr.Column():
            audio_output = gr.Audio(label="🎧 Audio Output", type="filepath", visible=False)
            file_output = gr.File(label="📄 Download Caption File", visible=False)

    # Interactions
    gen_btn.click(fn=generate_caption, inputs=image_input, outputs=caption_output)
    tts_btn.click(fn=speak_caption, inputs=caption_output, outputs=audio_output).then(
        lambda: gr.update(visible=True), outputs=audio_output
    )
    download_btn.click(fn=save_caption_file, inputs=caption_output, outputs=file_output).then(
        lambda: gr.update(visible=True), outputs=file_output
    )

demo.launch(share=True)
```
